```python
# ==========================================================
# Import Libraries
# ==========================================================

import os

import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
```

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
PROJECT_DIR = "/content/drive/MyDrive/AI_Brand_Project"

OUTPUT_DIR = os.path.join(PROJECT_DIR,"output")
```

```python
ranking_df = pd.read_csv(

os.path.join(
OUTPUT_DIR,
"brand_rankings.csv"
)

)

ranking_df.head()
```

```python
mention_rate = (

ranking_df

.groupby("Brand")

["Mention"]

.mean()

.rename("Mention_Rate")

)
```

```python
avg_rank = (

ranking_df

.groupby("Brand")

["Rank"]

.mean()

.rename("Average_Rank")

)
```

```python
# ==========================================================
# Top1 Rate
# ==========================================================

top1 = (
    ranking_df.assign(Top1=ranking_df["Rank"] == 1)
    .groupby("Brand")["Top1"]
    .mean()
    .rename("Top1_Rate")
)

top1
```

```python
# ==========================================================
# Top3 Rate
# ==========================================================

top3 = (
    ranking_df.assign(Top3=ranking_df["Rank"] <= 3)
    .groupby("Brand")["Top3"]
    .mean()
    .rename("Top3_Rate")
)

top3
```

```python
visibility = pd.concat(

[

mention_rate,

avg_rank,

top1,

top3

],

axis=1

)

visibility
```

```python
#Section 3
visibility["Rank_Score"]=1/visibility["Average_Rank"]
```

```python
scaler=MinMaxScaler()

cols=[

"Mention_Rate",

"Rank_Score",

"Top1_Rate",

"Top3_Rate"

]

visibility[

[

c+"_Norm"

for c in cols

]

]=scaler.fit_transform(

visibility[cols]

)
```

```python
visibility["AI_Visibility"]=(

visibility[

[

"Mention_Rate_Norm",

"Rank_Score_Norm",

"Top1_Rate_Norm",

"Top3_Rate_Norm"

]

]

.mean(axis=1)

*100

)
```

```python
visibility=visibility.sort_values(

"AI_Visibility",

ascending=False

)
```

```python
visibility
```

```python
visibility.to_csv(

os.path.join(

OUTPUT_DIR,

"AI_Visibility_Score.csv"

),

index=True

)

print("Saved!")
```
